```python
import pandas as pd
df = pd.read_csv("heart_failure_clinical_records_dataset.csv")

```

```python
pd.set_option("display.max_rows", None)
df["age_group"] = pd.cut(
    df["age"],
    bins=[40,50,60,70,80,90,100],
    right=False,
    labels=["40s","50s","60s","70s","80s","90s"]
)

```

```python
df.groupby("age_group")["DEATH_EVENT"].agg(
    individuals = "count",
    death_count = "sum",
    death_rate = "mean",
    
).assign(
    death_rate_changes= lambda x: (x["death_rate"].pct_change()*100).map(
        lambda x: f"+{x:.2f}%" if pd.notna(x) else "N/A"
    ),
    survival_rate = lambda y: (1 - y["death_rate"]).map(
        lambda y: f"{y:.2f}%" if pd.notna(y) else "N/A"
    )
)
# Shows mortality rate per decade of life and it's changes.
```
