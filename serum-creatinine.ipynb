```python
import pandas as pd
import numpy as np
df = pd.read_csv("heart_failure_clinical_records_dataset.csv")
```

```python
df["creatinine_quartile"] = pd.qcut(
    df["serum_creatinine"],
    q=4,
    labels=["Q1","Q2","Q3","Q4"]
)
df["creatinine_quartile"].value_counts().sort_index()
```

```python
result = df.groupby(
    "creatinine_quartile",
    observed=True
)["DEATH_EVENT"].agg(
    patients="count",
    death_count="sum",
    death_rate="mean"
).assign(
    death_rate_changes = lambda x: (x["death_rate"].pct_change()*100).map(
        lambda x: f"{x:.2f}%" if pd.notna(x) else "N/A"
    )
)

result
```

```python
# we can conclude that higher serum creatinine is generally associated with higher mortality rate
# though moving from second quartile to third quartile shows an slight decrease in death rate 
# and the reason behind that could be a shortage of data but it needs furthur academic
# reaserch to find a clear conclusion.
```
